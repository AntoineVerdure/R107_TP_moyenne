                                                                TP | Calcul des moyennes

.La situation

L'évaluation de l'année est réalisée par le calcul de la moyenne annuelle dans chacune des 3 UE formant le BUT 1.

-UE1 : dominante réseaux
-UE2 : dominante télécoms
-UE3 : dominante programmation
Une moyenne semestrielle est produite pour chaque ressource et chaque SAÉ.

La moyenne semestrielle pour chaque UE est une moyenne de toutes les notes du semestre, pondérée par des coefficients qui dépendent de l'UE et du semestre.

En fin d'année :

-Si la moyenne sur l'année de chacune des trois UE est ≥10
 : l'année est validée et le passage en BUT 2 est autorisé
 
-Si la moyenne sur l'année de 2 UE est ≥10
 alors que la moyenne de la troisième UE est comprise entre 8 (inclus) et 10 (exclu) : l'année de BUT 1 n'est pas validée mais le passage en BUT 2 est autorisé. Il faudra alors impérativement valider cette dernière UE en BUT 2 (au niveau 2, ce qui validera automatiquement le niveau 1 )
 
-Si la moyenne sur l'année de 2 UE est ≥10
 alors que la moyenne de la troisième UE est inférieure à 8 : l'année n'est pas validée et le passage en BUT 2 n'est pas autorisé. Un redoublement peut-être proposé à l'appréciation du conseil des enseignants.
 
-Si la moyenne sur l'année de 2 ou 3 UE est inférieure à 10 : l'année n'est pas validée et le passage en BUT est exclu.

Algo : 
-Demander les coeffs de chaques R/SAE dans l'UE 1
-Demander d'inscire les notes des R/SAE défini dans l'UE seléctionner
-Calculer la moyenne de l'UE1
-Afficher la moyenne de l'UE1
-Recommencer pour l'UE2 et l'UE3
-Afficher si l'UE est supérieur/égal à 10
        -> UE vous êtes dans le vert
    Sinon UE vous êtes dans le rouge
